1. What is unsupervised learning in the context of machine learning ?

    -> Unsupervised learning is a fundamental paradigm in machine learning where algorithms analyze unlabeled datasets to discover intrinsic structures, hidden patterns, or underlying data distributions without human guidance. Unlike supervised learning, which relies on predefined target outputs, unsupervised methods operate solely on input features $X = \{x_1, x_2, \dots, x_n\}$, aiming to model the underlying probability density $P(X)$ or learn compact representations of the data space.

    The paradigm encompasses four main theoretical subfields:
      * Clustering: Partitioning data into distinct groups based on similarity metrics, such as distance or density, so that intra-cluster variance is minimized while inter-cluster variance is maximized (e.g., K-Means, DBSCAN).
      * Dimensionality Reduction: Mapping high-dimensional vectors to a lower-dimensional subspace while preserving maximal variance or geometric topology, mitigating the curse of dimensionality (e.g., PCA, Autoencoders).
      * Density Estimation and Generative Modeling: Estimating the underlying probability distribution to assess point likelihoods or sample entirely new synthetic data points (e.g., Gaussian Mixture Models, VAEs).
      * Association Rule Learning: Identifying non-obvious conditional relationships and co-occurrences among feature sets within transactional data (e.g., Apriori algorithm).

2. How does K-Means clustering algorithm work ?

    **Initialization**

    * Define the number of clusters, $K$.
    * Select $K$ initial centroid positions $\{\mu_1, \mu_2, \dots, \mu_K\}$ randomly or using the **K-Means++** algorithm to ensure spread-out starting points.

    **Assignment Step**

    * Compute the Euclidean distance from every data point $x_i$ to each of the $K$ centroids:
    $$d(x_i, \mu_k) = \Vert{} x_i - \mu_k \Vert{}_2$$


    * Assign each data point $x_i$ to its closest centroid's cluster $S_k$:
    $$c^{(i)} = \arg\min_k \Vert{} x_i - \mu_k \Vert{}^2$$



    **Update Step**

   * Recalculate the position of each centroid $\mu_k$ by taking the arithmetic mean of all data points currently assigned to cluster $S_k$:

    $$\mu_k = \frac{1}{\vert{}S_k\vert{}} \sum_{x \in S_k} x$$

    **Termination Step**

    * Repeat the Assignment and Update steps iteratively until a convergence    condition is satisfied:
    * Centroid positions stop changing ($\Delta \mu \approx 0$).
    * Cluster assignments for all data points remain unchanged.
    * The maximum allocated number of iterations is reached.

3. Explain the concept of a dendrogram in hierarchical clustering.

    ## Dendrogram in Hierarchical Clustering

    A **dendrogram** is a tree-like diagram used in **hierarchical clustering** to visually represent how data points or clusters are grouped together based on their similarity or distance.

    In a dendrogram, **each leaf represents an individual data point**, while the branches show the process of merging similar data points into larger clusters. The **height of a branch** indicates the distance or dissimilarity between the clusters being merged. A smaller height means the clusters are more similar, while a larger height indicates greater dissimilarity.

    Hierarchical clustering generally follows two approaches:

    1. **Agglomerative clustering:** Starts with each data point as a separate cluster and progressively merges the closest clusters.
    2. **Divisive clustering:** Starts with all data points in one cluster and progressively divides them into smaller clusters.

    A horizontal line can be drawn across the dendrogram at a selected distance level to determine the required number of clusters.

        **Example:** If a dendrogram is cut at a particular height and it produces three separate branches, the dataset can be considered to have **three clusters**.

    Thus, a dendrogram helps visualize the **hierarchical relationships and similarity among data points** and helps determine an appropriate number of clusters.


4. What is the main difference between K-Means and Hierarchical Clustering ?

    **Difference Between K-Means and Hierarchical Clustering**
* K-Means Clustering
    * It is a partition-based clustering technique.
    * The number of clusters (K) must usually be specified in advance.
    * It assigns data points to clusters based on their distance from the cluster centroid.
    * It repeatedly updates the centroids until the clusters become stable.
    * It is generally faster and more efficient for large datasets.
    * The final output is a set of separate clusters.
    * It is sensitive to initial centroid selection and outliers.

* Hierarchical Clustering
    * It is a hierarchy-based clustering technique.
    * The number of clusters does not need to be specified initially.
    * It groups data points based on their similarity or distance.
    * It creates a tree-like structure called a dendrogram.
    * It is generally more computationally expensive than K-Means.
    * It is suitable for understanding relationships between different clusters.
    * The desired number of clusters can be selected by cutting the dendrogram at a particular level.

5. What are the advantages of DBSCAN over K-Means ?

    #### Advantages of DBSCAN over K-Means
    #### DBSCAN (Density-Based Spatial Clustering of Applications with Noise)


    1. **No need to specify the number of clusters:**
      DBSCAN does not require the number of clusters to be defined in advance, unlike K-Means.

    2. **Detects clusters of arbitrary shapes:**
      DBSCAN can identify clusters with irregular or complex shapes, whereas K-Means generally works best with spherical or circular clusters.

    3. **Handles noise and outliers:**
      DBSCAN can identify isolated data points as **noise or outliers** instead of forcing them into a cluster.

    4. **Less sensitive to initial conditions:**
      Unlike K-Means, DBSCAN does not depend on randomly selected initial cluster centroids.

    5. **Works well with unevenly shaped data:**
      It can discover naturally occurring dense regions even when clusters are not uniformly shaped.

    6. **Useful for spatial data:**
      DBSCAN is particularly effective for geographical and location-based datasets where points form dense regions.

    **The major advantage of DBSCAN over K-Means is that it can find clusters of arbitrary shapes, detect outliers, and does not require the number of clusters to be specified beforehand.**

    2. **To select the optimal number of clusters:**

    It can be used to compare clustering performance across different algorithms or parameter settings.

     * Around 0: Overlapping or poorly separated clusters.
     * Negative value: Data points may have been assigned to the wrong clusters.

9. How does DBSCAN identify noise points .

    DBSCAN (Density-Based Spatial Clustering of Applications with Noise) identifies noise points by examining the density of data points around each point.

    ### Process of Identifying Noise
  * Defines Two Parameters:

    DBSCAN uses two main parameters:

        * ε (Epsilon): The maximum distance considered for neighboring points.
        * MinPts: The minimum number of points required within the ε-neighborhood to form a dense region.
  * **Checks the Neighborhood:**
For each data point, DBSCAN counts how many other points are located within its ε distance.
 * **Identifies Core Points:**
If a point has at least MinPts points within its ε-neighborhood, it is considered a core point.
 * **Identifies Border Points:**
A point that does not have enough neighbors to be a core point but lies within the ε-neighborhood of a core point is called a border point.
 * **Identifies Noise Points:**
If a point is neither a core point nor a border point, it is classified as a noise (outlier) point.
Example

    If ε = 2 and MinPts = 5, a point must have at least 5 nearby points within a distance of 2 to be considered a core point. An isolated point with very few nearby neighbors will be classified as noise.

10. Define inertia in the context of K-Means .

     ### Inertia in K-Means Clustering

    Inertia is a measure used in K-Means clustering to evaluate how closely the data points are grouped around their respective cluster centers (centroids).

    ### Key Points

    1. **Definition:**
    Inertia is the sum of the squared distances between each data point and the centroid of the cluster to which it belongs.

    2. **Purpose:**
    It measures the compactness of clusters. Lower inertia means that data points are closer to their cluster centroids.

    3. **Calculation:**
    It can be represented as:

    $$ \text{Inertia} = \sum_{i=1}^{n} ||x_i - \mu_{c_i}||^2 $$

    Where:

    * \(x_i\) = individual data point
    * \(\mu_{c_i}\) = centroid of the cluster containing the data point
    * \(n\) = total number of data points
    4. **Relation to K:**
    As the number of clusters (K) increases, inertia generally decreases because data points have more centroids available.
    5. **Elbow Method:**
    Inertia is commonly used with the Elbow Method to select a suitable value of K. The point where adding more clusters provides only a small reduction in inertia is called the elbow point.

11. What is the elbow method in K-Means clustering ?

    ### Elbow Method in K-Means Clustering

      The Elbow Method is a technique used in K-Means clustering to determine the optimal number of clusters (K) for a dataset.

    #### **How the Elbow Method Works**

    1. **Select different values of K:**
    Run the K-Means algorithm for different numbers of clusters, such as K = 1, 2, 3, 4, 5, ....

    2. **Calculate Inertia:**
    For each value of K, calculate the inertia, which represents the sum of squared distances between data points and their cluster centroids.

    3. **Plot the Results:**
    Plot K on the X-axis and inertia on the Y-axis.
    
    4. **Find the Elbow Point:**
    Inertia decreases as K increases. Initially, the decrease is large, but after a certain point, the improvement becomes much smaller. This point creates a shape similar to an elbow.
    5. **Select K:**
    The K value at the elbow is generally selected as the optimal number of clusters because adding more clusters after this point provides relatively little improvement.

12. Describe the concept of "density" in DBSCAN .

    ### Density in DBSCAN

    In DBSCAN (Density-Based Spatial Clustering of Applications with Noise), density refers to the number of data points located within a specified neighborhood around a particular data point.

    Key Points

* **Epsilon (ε):**
DBSCAN defines a radius around each data point. This radius is called ε (Epsilon).
* **MinPts:**
It specifies the minimum number of data points that should be present within the ε-neighborhood for the area to be considered sufficiently dense.
* **Dense Region:**
If a point has at least MinPts points within its ε distance, it is considered a core point, indicating a dense region.
* **Sparse Region:**
If there are fewer than MinPts points within the ε-neighborhood, the region is considered sparse.
* **Cluster Formation:**
DBSCAN connects nearby dense regions to form clusters. Therefore, clusters are essentially continuous areas of high data-point density.
* **Noise Detection:**
Points located in sparse regions and not connected to any dense region can be classified as noise or outliers.


13. Can hierarchical clustering be used on categorical data ?

    Yes, hierarchical clustering can be used on categorical data, but an appropriate distance or similarity measure must be used instead of directly applying Euclidean distance.

    ## Key Points
    1. **Categorical Data:**

  * **Gender**: Male/Female
 * **Department**: HR/IT/Finance
 * **City**: Delhi/Mumbai/Chennai

15. Explain the term "linkage criteria" in hierarchical clustering .

    **Linkage Criteria in Hierarchical Clustering**

    Linkage criteria is the method used in hierarchical clustering to determine the distance or dissimilarity between two clusters. It decides which clusters should be merged during the clustering process.

    **Types of Linkage Criteria**

    1. Single Linkage:
    Measures the distance between the closest pair of points from two clusters.
        * Can create long, chain-like clusters.
        * Sensitive to noise.
    2. Complete Linkage:
    Measures the distance between the farthest pair of points from two clusters.
        * Produces compact and well-separated clusters.
    3. Average Linkage:
    Calculates the average distance between all pairs of points from two clusters.
        * Provides a balance between single and complete linkage.
    4. Ward's Linkage:
    Merges clusters that result in the smallest increase in within-cluster variance.
       * Often produces compact and relatively equal-sized clusters.

16. Why might K-Means clustering perform poorly on data with varying cluster sizes or densities ?

    K-Means clustering may perform poorly when clusters have different sizes or densities because it assumes that clusters are generally compact, spherical, and of similar size.

    ## Key Points

     1. **Assumes Similar Cluster Sizes:** K-Means tends to divide data based on distance from centroids. If one cluster is much larger than another, the larger cluster can influence the centroid and affect the smaller cluster's assignments.
    2. **Assumes Similar Density:** K-Means does not explicitly consider the density of data points. Dense and sparse clusters may therefore be incorrectly grouped.
    3. **Sensitive to Cluster Shape:** K-Means works best with roughly spherical or circular clusters. It may struggle with irregularly shaped clusters.
    4. **Centroid-Based Approach:** Each point is assigned to the nearest centroid. This can cause points from different-density clusters to be incorrectly assigned.
    5. **Unequal Cluster Sizes:** A small but dense cluster may be merged with a nearby larger cluster because the distance-based objective favors minimizing overall squared distances.

    Example :  Suppose a dataset contains:

    * One large, widely spread cluster
    * One small, highly dense cluster

    K-Means may incorrectly divide the large cluster or assign some points from the small cluster to the large cluster because it tries to minimize the overall distance to the centroids.

17. What are the core parameters in DBSCAN, and how do they influence clustering ?

    ## Core Parameters in DBSCAN :
    DBSCAN (Density-Based Spatial Clustering of Applications with Noise) mainly uses two important parameters: ε (Epsilon) and MinPts. These parameters determine how DBSCAN identifies dense regions, forms clusters, and detects noise.

    1. Epsilon (ε)

    * ε defines the maximum distance around a data point for finding its neighboring points.
    * A larger ε creates a larger neighborhood, which can result in larger or fewer clusters.
    * A smaller ε creates a smaller neighborhood, which may produce more clusters and classify more points as noise.
    Choosing an appropriate ε is important for obtaining meaningful clusters.
    2. MinPts
    * MinPts specifies the minimum number of points required within the ε-neighborhood to consider a point a core point.
    * A higher MinPts requires a denser region to form a cluster.
    * A lower MinPts allows less-dense regions to become clusters and may increase the effect of noise.

18. How does K-Means++ improve upon standard K-Means initialization ?

    K-Means++ is an improved method for selecting the initial cluster centroids in the K-Means clustering algorithm. Standard K-Means often selects initial centroids randomly, which can lead to poor clustering results.

    ## Key Points

    1. **Better Initial Centroids:**
    K-Means++ selects initial centroids more carefully instead of choosing all of them randomly.
    2. **Spreads Centroids:**
    After selecting the first centroid randomly, each subsequent centroid is chosen with a higher probability if it is farther from the already selected centroids.
    3. **Reduces Poor Initialization:**
    Standard K-Means can select multiple initial centroids close to each other. K-Means++ reduces this problem by spreading the centroids across the dataset.
    4. **Improves Clustering Quality:**
    Better initial centroids usually lead to more compact and well-separated clusters.
    5. **Faster Convergence:**
    Because the starting centroids are better positioned, K-Means++ often requires fewer iterations to reach a stable solution.
    6. **More Consistent Results:**
    It reduces the likelihood of obtaining a poor local solution caused by an unlucky random initialization.

19. What is agglomerative clustering ?

    # Agglomerative Clustering

    Agglomerative clustering is a type of hierarchical clustering that follows a bottom-up approach. It starts by treating every data point as an individual cluster and then progressively merges the most similar or closest clusters until all data points form a single cluster or the desired number of clusters is obtained.

    ### Agglomerative Clustering Works
    1. **Start with Individual Clusters:**
    Each data point is initially considered a separate cluster.
    2. **Calculate Distances:**
    The distance or similarity between the clusters is calculated using a selected distance metric and linkage criterion.
    3. **Merge Closest Clusters:**
    The two most similar clusters are merged into one cluster.
    4. **Repeat the Process:**
    The distance between the new clusters is recalculated, and the closest clusters are merged again.
    5. **Create a Dendrogram:**
    The complete merging process is represented using a dendrogram, which shows the hierarchical relationship between the data points.
    6. **Select the Number of Clusters:**
    The dendrogram can be cut at a suitable level to obtain the required number of clusters.

20. What makes Silhouette Score a better metric than just inertia for model evaluation ?

    Silhouette Score can provide a more informative evaluation of clustering than inertia alone because it considers both cluster compactness and cluster separation.

    ### Key Points

    1. **Inertia Measures Only Compactness:**
    Inertia measures the sum of squared distances between data points and their cluster centroids. Lower inertia means more compact clusters.
    2. **Inertia Always Decreases with More Clusters:**
    As the number of clusters (K) increases, inertia generally decreases. Therefore, a lower inertia does not necessarily mean that the clustering is better.
    3. **Silhouette Score Measures Two Factors:**

        It considers:
          * How close a point is to other points in its own cluster.
          * How far it is from points in the nearest neighboring cluster.
    4. **Evaluates Cluster Separation:**
    A high Silhouette Score indicates that clusters are both well-formed and well-separated.
    5. **Easier Comparison:**
    The Silhouette Score ranges from -1 to +1, making it easier to compare clustering results across different values of K.
    6. **Identifies Poor Assignments:**
    A negative Silhouette Score can indicate that a data point may have been assigned to the wrong cluster.

```python
# 21. Generate synthetic data with 4 centers using make_blobs and apply K-Means clustering. Visualize using a scatter plot .

import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, y = make_blobs(
    n_samples=400,
    centers=4,
    cluster_std=1.0,
    random_state=42
)

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

y_kmeans = kmeans.fit_predict(X)

plt.figure(figsize=(8, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y_kmeans,
    cmap="viridis",
    s=50,
    alpha=0.7
)
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    c="red",
    marker="X",
    s=200,
    label="Centroids"
)
plt.title("K-Means Clustering with 4 Centers")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

# Display the plot
plt.show()
```

```python
# 23. Generate synthetic data using make_moons and apply DBSCAN. Highlight outliers in the plot .

import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN

X, y = make_moons(
    n_samples=300,
    noise=0.08,
    random_state=42
)

dbscan = DBSCAN(
    eps=0.2,
    min_samples=5
)
labels = dbscan.fit_predict(X)
outliers = labels == -1

plt.figure(figsize=(8, 6))

plt.scatter(
    X[~outliers, 0],
    X[~outliers, 1],
    c=labels[~outliers],
    cmap="viridis",
    s=50,
    alpha=0.7,
    label="Clusters"
)
plt.scatter(
    X[outliers, 0],
    X[outliers, 1],
    c="red",
    marker="x",
    s=100,
    linewidths=2,
    label="Outliers"
)

plt.title("DBSCAN Clustering on Make-Moons Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
print("Number of outliers:", outliers.sum())
```

```python
# 24. Load the Wine dataset and apply K-Means clustering after standardizing the features. Print the size of each cluster .

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import numpy as np

wine = load_wine()
X = wine.data

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)
labels = kmeans.fit_predict(X_scaled)
cluster_sizes = np.bincount(labels)

print("Size of each cluster:")
for cluster, size in enumerate(cluster_sizes):
    print(f"Cluster {cluster}: {size} data points")
```

```python
# 29.  Create synthetic data using make_blobs and evaluate silhouette scores for k = 2 to 5. Display as a bar chart .

import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
X, y = make_blobs(
    n_samples=500,
    centers=4,
    cluster_std=1.0,
    random_state=42
)
k_values = range(2, 6)
silhouette_scores = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X)
    score = silhouette_score(X, labels)
    silhouette_scores.append(score)

for k, score in zip(k_values, silhouette_scores):
    print(f"K = {k}, Silhouette Score = {score:.4f}")

plt.figure(figsize=(8, 5))
plt.bar(k_values, silhouette_scores)

plt.title("Silhouette Scores for Different Values of K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.xticks(list(k_values))
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
# 30. Load the Iris dataset and use hierarchical clustering to group data. Plot a dendrogram with average linkage .

import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from scipy.cluster.hierarchy import dendrogram, linkage

iris = load_iris()
X = iris.data

Z = linkage(X, method="average", metric="euclidean")

plt.figure(figsize=(12, 6))

dendrogram(
    Z,
    labels=iris.target,
    leaf_rotation=90,
    leaf_font_size=8
)

plt.title("Hierarchical Clustering Dendrogram - Average Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
# 31. Generate synthetic data with overlapping clusters using make_blobs, then apply K-Means and visualize with decision boundaries.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, y = make_blobs(
    n_samples=500,
    centers=3,
    cluster_std=2.5,
    random_state=42
)

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

labels = kmeans.fit_predict(X)
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 500),
    np.linspace(y_min, y_max, 500)
)

grid_points = np.c_[xx.ravel(), yy.ravel()]
Z = kmeans.predict(grid_points)
Z = Z.reshape(xx.shape)

plt.figure(figsize=(10, 7))

plt.contourf(
    xx, yy, Z,
    alpha=0.2,
    cmap="viridis"
)

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    cmap="viridis",
    s=40,
    edgecolor="black",
    alpha=0.7
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker="X",
    s=250,
    c="red",
    edgecolor="black",
    label="Centroids"
)
plt.contour(
    xx, yy, Z,
    levels=np.arange(0.5, kmeans.n_clusters, 1),
    colors="black",
    linewidths=1.5
)

plt.title("K-Means Clustering with Decision Boundaries")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
# 34. Load the Breast Cancer dataset and compare inertia values for K = 2 to 6 using K-Means. Show results in a line plot .

import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

data = load_breast_cancer()
X = data.data

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

k_values = range(2, 7)
inertia_values = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    inertia_values.append(kmeans.inertia_)
for k, inertia in zip(k_values, inertia_values):
    print(f"K = {k}, Inertia = {inertia:.2f}")

plt.figure(figsize=(8, 5))
plt.plot(
    k_values,
    inertia_values,
    marker="o",
    linewidth=2
)

plt.title("K-Means Inertia for Breast Cancer Dataset")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(list(k_values))
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
# 40. Load the Digits dataset, apply PCA to reduce to 3 components, then use KMeans and visualize with a 3D scatter plot .

import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

digits = load_digits()
X = digits.data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)

kmeans = KMeans(
    n_clusters=10,
    random_state=42,
    n_init=10
)

labels = kmeans.fit_predict(X_pca)
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

scatter = ax.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    X_pca[:, 2],
    c=labels,
    cmap="viridis",
    s=20,
    alpha=0.7
)
centers = kmeans.cluster_centers_

ax.scatter(
    centers[:, 0],
    centers[:, 1],
    centers[:, 2],
    marker="X",
    c="red",
    s=200,
    edgecolor="black",
    linewidth=2,
    label="Cluster Centers"
)

ax.set_title("K-Means Clustering of Digits Dataset using 3D PCA")
ax.set_xlabel("Principal Component 1")
ax.set_ylabel("Principal Component 2")
ax.set_zlabel("Principal Component 3")

ax.legend()
plt.show()
```
