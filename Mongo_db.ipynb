# Theoretical Questions

1. What are the key differences between SQL and NoSQL databases ?
    ### SQL Databases (Relational)

* Data Structure: Stores data in structured tables composed of fixed rows and columns.
* Schema Flexibility: Enforces a rigid, predefined schema that must be explicitly declared before data insertion.
* Scaling Mechanism: Scales primarily vertically by adding CPU, RAM, or VRAM capacity to a single database server.
* Transaction Reliability: Guarantees strict ACID (Atomicity, Consistency, Isolation, Durability) compliance for transaction safety.
* Query Standard: Interrogates and manipulates data using standardized Structured Query Language (SQL).
* Data Relationships: Handles complex relational JOIN operations across multiple tables with high efficiency.
* Primary Examples: PostgreSQL, MySQL, Oracle DB, Microsoft SQL Server.
### NoSQL Databases (Non-Relational)

* Data Structure: Stores data using dynamic structures like JSON documents, key-value pairs, wide-column stores, or node-edge graphs.
* Schema Flexibility: Features dynamic, flexible schemas that accept unstructured or semi-structured data without upfront definitions.
* Scaling Mechanism: Scales horizontally by partitioning and sharding datasets across distributed server clusters.
* Transaction Reliability: Prioritizes BASE properties (Basically Available, Soft-state, Eventual consistency) for high availability and low-latency throughput.
* Query Standard: Interrogates data using specialized APIs, JSON-based query constructs, or object-oriented drivers.
* Data Relationships: Avoids relational JOIN operations; relies on denormalized data models or localized document lookups.
* Primary Examples: MongoDB, Apache Cassandra, Redis, Amazon DynamoDB, Neo4j.

2. What makes MongoDB a good choice for modern applications ?

    -> MongoDB has become a cornerstone database for modern applications due to its flexibility, developer-centric design, and horizontal scaling capabilities.

* **Flexible, Dynamic Schema:** Documents in the same collection do not need to share the exact same structure. Fields can be added on the fly, making it easy to iterate rapidly and continuously deploy features without running locking database migrations.

* **Horizontal Scalability (Sharding):** Native support for auto-sharding enables datasets and query workloads to be partitioned automatically across clusters of servers, supporting massive write throughput and high storage demands.

* **Built-In High Availability (Replica Sets):** Automatically manages database redundancy through multi-node replica sets. It provides automatic failover and primary election if a active node goes down.

* **Rich Querying & Aggregation Framework:** Supports complex indexing (including geospatial, full-text search, and compound indexes), nested document queries, and multi-stage aggregation pipelines for real-time analytics.

* **ACID Transactions Support:** Supports multi-document ACID transactions across single and distributed clusters, providing relational-level data integrity guarantees when strictly necessary.

* **Developer Experience & Ecosystem:** Deep integration with popular runtime ecosystems (e.g., Node.js, Python, Java) and modern cloud platform management through MongoDB Atlas.

3. Explain the concept of collections in MongoDB .

    -> In MongoDB, a collection is a grouped gathering of stored BSON (Binary JSON) documents. It is the MongoDB equivalent of a table in a traditional relational database (SQL), but with key structural differences designed for flexibility.
* **No Predefined Schema (Schema-Flexible):** Unlike SQL tables, collections do not enforce a fixed structure. Documents inside the same collection can have different fields, varied data types, or distinct sub-document hierarchies.

* **Document Enclosure:** A collection holds individual documents (records), and each document within a collection automatically receives a unique primary key identifier field named _id.

* **On-Demand Creation:** Collections are created automatically when the first document is inserted into them—there is no strict requirement to run an explicit CREATE TABLE command beforehand.

* **Indexing Support:** Indexes can be created directly on specific fields within a collection (including embedded fields and arrays) to accelerate search and aggregation performance.

4. How does MongoDB ensure high availability using replication ?

    -> MongoDB ensures high availability through Replica Sets—a cluster of connected MongoDB instances that maintain the same dataset through continuous replication and automated failover.

* **Automatic Failover & Elections:** If the Primary node goes offline or stops heartbeating (default 10-second timeout), the remaining Secondaries hold an automated election using a Raft-like consensus protocol to select a new Primary without human intervention.

* **Oplog Replication Pipeline:** Writes applied to the Primary are immediately recorded as idempotent actions in its capped local.oplog.rs collection. Secondaries continuously tail this log and apply the operations locally.

* **Configurable Read Preferences:** Applications can offload read operations from the Primary to Secondaries (e.g., secondaryPreferred or nearest) to handle heavy read traffic across geographical locations.

* **Tunable Write Concern:** Controls the level of acknowledgment requested from MongoDB for write operations. Using w: "majority" ensures writes are committed to a majority of replica set members before acknowledging success, preventing rollback during sudden failovers.

6. What is the role of indexes in MongoDB, and how do they improve performance ?

    -> An index in MongoDB is a specialized data structure (typically a B-tree) that stores a small, ordered portion of a collection’s dataset in memory. It maps specific field values directly to the physical location of their corresponding documents on disk.

    Without indexes, MongoDB must perform a Collection Scan (COLLSCAN), examining every single document in a collection to locate results matching a query. Indexes allow MongoDB to execute an Index Scan (IXSCAN), drastically reducing disk I/O, latency, and CPU overhead.

## Indexes Improve Performance
* **Eliminating Full Collection Scans:** Converts linear O(N) document scans into logarithmic O(log N) tree lookups, reducing queried documents from millions to a handful.
* **Optimizing In-Memory Sorting (SORT operations):** Because index keys are stored pre-sorted, MongoDB can read sorted documents directly from the index. Without an index, MongoDB must load all matching records into RAM to perform an in-memory sort, risking memory limit errors.
* **Enabling Covered Queries:** If a query requests only fields contained within an index, MongoDB serves the request directly from RAM (Covered Query) without fetching data from disk at all.
* **Enforcing Unique Constraints:** Unique indexes prevent duplicate entries (e.g., duplicate email addresses) directly at the database engine level with zero application-side overhead.
* **Accelerating Range Searches:** Indexes allow range operators ($gt, $lt, $gte, $lte) to seek to a specific key and read only the relevant bounded segment.

### Common Index Types in MongoDB

* **Single Field Index:** Indexes a single field in ascending (1) or descending (-1) order. (MongoDB automatically creates a default single-field index on _id).

* **Compound Index:** Indexes multiple fields together. Follows the ESR Rule (Equality, Sort, Range) for field ordering.

* **Multikey Index:** Automatically created when indexing fields containing array values, indexing each array element separately.

* **Text & Vector Indexes:** Supports full-text search operators ($text) and vector embeddings (via MongoDB Atlas Vector Search) for AI/RAG applications.

* **Geospatial Indexes:** Facilitates distance and location queries (2dsphere / 2d).

7. Describe the stages of the MongoDB aggregation pipeline .

    -> The MongoDB aggregation pipeline processes documents sequentially through specialized transformation stages, passing each stage's output as the next stage's input.

* **Filtering & Selection:** 'match' filters documents matching specific criteria; $sample randomly selects a designated subset of documents.

* **Reshaping & Projection:** 'project' modifies field structures; addFields/set append or update fields; unset removes fields; replaceRoot/ replaceWith replace the document's top-level structure.

* **Grouping & Aggregation:** group aggregates documents by an _id key with functions ; unwind flattens array elements into individual documents; bucket/bucketAuto organize documents into numerical or automated quantile ranges.

* **Joins & Relationships:** lookup performs left outer joins across collections; graphLookup executes recursive structural searches on hierarchical data.

* **Sorting & Pagination:** sort orders documents by key; limit caps the total document output; skip omits a specified number of initial documents.

* **Search Execution:** search executes full-text search against Lucene indexes; vectorSearch performs Approximate Nearest Neighbor (ANN) search on embeddings.

* **Output & Writing:** out replaces a target collection with pipeline results; merge writes results to a target collection with customized insert/update logic.

8. What is sharding in MongoDB? How does it differ from replication ?

    -> MongoDB sharding is a horizontal scaling strategy designed to handle massive data volumes and high-throughput write operations by partitioning datasets across a cluster of servers. Instead of expanding a single database instance vertically with more RAM or CPU, sharding distributes collection data across multiple physical shards based on an indexed field called the shard key. Each shard acts as an independent storage unit holding a distinct subset of the overall data, while background routing processes transparently direct incoming read and write requests to the appropriate target shard. The system dynamically monitors data distribution, automatically splitting full data blocks into smaller chunks and rebalancing them across shards to prevent hardware bottlenecks and maintain consistent performance.

    Sharding differs fundamentally from replication in both its core objectives and underlying operational mechanics. While replication creates redundant copies of the complete dataset across multiple nodes within a single replica set to provide high availability and failover protection, sharding splits the dataset so that each shard stores a unique fragment. In a standard replica set, a single primary node bears the burden of all incoming write operations while secondary nodes continuously sync from its oplog. In contrast, sharding establishes multiple primary nodes—one for each shard—which distributes write workloads horizontally across the cluster. To ensure fault tolerance alongside scalability, modern production deployments combine both concepts by deploying each individual shard as a standalone replica set.

### Replication in MongoDB
* **Primary Purpose:** Delivers high availability, fault tolerance, and data redundancy.

* **Data Distribution:** Every member node in a replica set maintains a complete, identical copy of the full dataset.

* **Write Management:** Directs all write operations exclusively to a single designated Primary node, which logs changes to its oplog for Secondaries to replicate.

* **Read Scalability:** Allows applications to offload read operations to Secondary nodes, distributing read traffic across the cluster.

* **Fault Recovery:** Elects a new Primary node automatically via a Raft-like consensus election if the active Primary fails or loses connectivity.

* **Hardware Footprint:** Requires a minimum of three nodes (such as a Primary and two Secondaries, or a Primary, Secondary, and Arbiter) to form a functional voting cluster.

### Sharding in MongoDB

* **Primary Purpose:** Delivers horizontal scaling (scale-out) for datasets and write workloads that exceed single-server hardware limits.

* **Data Distribution:** Partitions and splits a collection dataset into smaller, non-overlapping subsets distributed across multiple independent shards.

* **Write Management:** Distributes write workloads horizontally across multiple Primary nodes (one Primary per shard) simultaneously.

* **Routing System:** Routes client read and write operations to the correct shard using stateless mongos query routers and metadata stored on dedicated Config Servers.

* **Automatic Balancing:** Monitors data chunk sizes continuously, automatically splitting growing chunks and migrating them across shards to prevent hardware hotspots.

* **Hardware Footprint:** Requires a multi-tier infrastructure consisting of Config Servers, mongos routers, and multiple shards (where each shard is typically deployed as its own replica set for high availability).

10. What are the ACID properties in the context of MongoDB transactions ?

    -> In MongoDB, single-document updates are inherently atomic. However, for operations spanning multiple documents, collections, or databases, MongoDB uses sessions (ClientSession) to execute multi-document transactions that adhere strictly to ACID properties.

### Atomicity
* Concept: Guaranteed "all-or-nothing" execution.

* MongoDB Execution: Every read and write operation within a transaction is treated as a single unit. If any individual operation within a session.startTransaction() block fails, or if an explicit session.abortTransaction() is called, MongoDB rolls back all pending changes made during that session. No partial updates are ever written to the collection.

### Consistency
* Concept: Ensures data transitions from one valid database state to another while respecting operational rules and constraints.

* MongoDB Execution: MongoDB maintains internal invariants and enforces schema validation rules (JSON Schema constraints) configured on collections. If a transaction violates a validation rule or index constraint (such as a unique index violation), the entire transaction aborts, leaving the database state intact as it was before the transaction began.

### Isolation
* Concept: Prevents concurrent transactions from interfering with or observing each other's intermediate updates.

* MongoDB Execution: MongoDB provides Snapshot Isolation. When a transaction starts, it receives a point-in-time snapshot of the dataset. Operations inside the transaction read from this snapshot, and uncommitted updates remain completely invisible to external operations outside the session. If two concurrent transactions attempt to modify the exact same document, a write conflict occurs, causing one of the transactions to abort and require an application-side retry.

### Durability
* Concept: Guarantees that once a transaction commits, its mutated data survives system crashes and network outages.

* MongoDB Execution: Durability depends on the configured Write Concern and Read Concern. By default, committed multi-document transactions use a write concern of w: "majority". This ensures that all write operations in the transaction are committed to the WiredTiger journal and acknowledged by a majority of replica set nodes before returning success to the application, making committed updates permanent.

11. What is the purpose of MongoDB’s explain() function ?

    -> The primary purpose of MongoDB's explain() function is to inspect and analyze the query execution plan, allowing developers and DBAs to evaluate database performance, optimize query efficiency, and verify index usage.

    -> Instead of just returning dataset results, explain() exposes the inner mechanics of how the MongoDB query optimizer processes operational requests (find, aggregate, update, delete, etc.).

### Core Objectives of explain()
* **Detect Unindexed Scans:** Identifies whether a query performs an efficient Index Scan (IXSCAN) or a costly, unindexed Collection Scan (COLLSCAN).

* **Evaluate Index Efficiency:** Measures the ratio between the number of documents examined (totalDocsExamined) or index keys scanned (totalKeysExamined) versus the number of actual matching documents returned (nReturned).

* **Analyze Execution Time:** Provides precise execution timing metrics (executionTimeMillis) to isolate slow-running database operations.

* **Inspect Winning vs. Rejected Plans:** Reveals how the MongoDB query optimizer evaluated multiple candidate execution plans and chose the winning execution path.

### Verbosity Modes :
####explain() operates under three main verbosity modes:

* **queryPlanner (Default):** Runs the query optimizer to select the winning execution plan without executing the query itself.

* **executionStats:** Executes the winning query plan to completion and returns actual execution performance metrics (execution time, document scan counts).

* **allPlansExecution:** Runs the winning plan to completion while capturing candidate selection metrics for all evaluated/rejected plans.

12. How does MongoDB handle schema validation ?

    -> MongoDB enforces Schema Validation on collections using standard JSON Schema definitions, allowing dynamic, document-oriented storage while maintaining strict structural guardrails for core fields.

### Core Validation Mechanics

* **JSON Schema Standards:** Validation rules are declared via the $jsonSchema operator within the validator collection option. Rules can define required fields, specify data types (e.g., string, int, objectId), restrict value ranges, and enforce regex patterns.

* **Write Interception:** Validation triggers automatically during insert and update operations. If an incoming document or update violates the specified rules, MongoDB rejects the operation (by default) before writing to disk.

* **Document-Level Scope:** Validation applies to individual collection documents. Modifying validation rules for a collection does not automatically mutate or retroactively check existing legacy documents unless those legacy records are subsequently updated.

13. What is the difference between a primary and a secondary node in a replica set ?

### Primary Node

* **Write Authority:** Serves as the single authoritative entry point that receives and processes all client write operations (insert, update, delete).

* **Singleton Constraint:** Restricted to strictly one active Primary node per replica set at any given time to avoid split-brain data conflicts.

* **Oplog Generation:** Records every dataset mutation as an idempotent operation in its local capped local.oplog.rs collection.

* **Default Read Target:** Serves all application read requests by default (primary read preference) to guarantee immediate read-after-write consistency.

* **Step-Down Trigger:** Automatically steps down to a Secondary node if it loses connection to a majority quorum of the replica set members.

### Secondary Node

* **Data Synchronization:** Continuously tails the Primary's oplog and applies those operations locally to maintain an identical copy of the dataset.

* **Read-Only Enforced:** Rejects direct write operations from client applications; accepts write logs exclusively via internal replication.

* **Scalable Read Target:** Offloads read-heavy operations or analytics queries from the Primary when configured with appropriate read preferences .

* **Failover Eligibility:** Participates in automated Raft-like consensus elections to vote for and potentially step up as the new Primary if the active Primary fails.

* **Specialized Role Flexibility:** Can be configured into dedicated operational roles, such as Hidden (invisible to client applications for reporting), Delayed (lagged behind Primary to recover from accidental deletions), or Priority 0 (eligible to vote but barred from becoming Primary).

14. What security mechanisms does MongoDB provide for data protection ?

    MongoDB provides a multi-layered security model designed to safeguard sensitive data across access control, encryption, network isolation, and auditing.  

    ### Authentication (Identity Verification)

    Ensures that only verified clients and applications can establish a connection to the database.

    * **SCRAM (Salted Challenge Response Authentication Mechanism):** The default username/password mechanism using SHA-256 algorithm hashing.  
    * **x.509 Certificate Authentication:** Uses Client PKI certificates for secure, key-based authentication over TLS.  
    *  **Enterprise & Cloud Identity Integrations:** Integrates with centralized identity providers using LDAP, Kerberos, and OIDC / OAuth 2.0 (e.g., Okta, Azure AD) for Single Sign-On (SSO).  

    ### Authorization & Access Control (RBAC)

    Determines what specific database resources and operations a verified user can access.  

    * **Role-Based Access Control (RBAC):** Restricts privileges according to predefined roles (e.g., read, readWrite, dbAdmin, userAdmin).  
    * **Custom Roles:** Allows administrators to construct fine-grained privilege sets granting specific CRUD permissions down to individual collections or action levels.
    * **Principle of Least Privilege:** Enforces strict operational boundaries so applications only possess the minimum required access.  
    ### Comprehensive Data Encryption

    Protects sensitive data across all stages of its lifecycle:
    * **In-Transit Encryption (TLS/SSL):** Encrypts all network traffic between clients, application drivers, and database instances to prevent eavesdropping and man-in-the-middle attacks.  
    * **At-Rest Encryption:** Encrypts raw storage files on disk using AES-256 via the WiredTiger engine (supports customer-managed keys integrated with AWS KMS, Azure Key Vault, or GCP KMS).  
    * **Client-Side Field Level Encryption (CSFLE) & Queryable Encryption:**
        * **In-Use Security:** Encrypts specific sensitive fields (e.g., SSNs, credit card numbers, legal contracts) before sending data over the wire.  
        * **Zero-Trust Storage:** The database engine only sees encrypted ciphertext; decryption keys remain exclusively with the application client.  
        * **Queryable Encryption:** Allows searching and querying encrypted data directly without decrypting it on the server.  
    ### **Network Security & Isolation**

    Prevents unauthorized public or untrusted network access to cluster infrastructure:  
    * **IP Access Lists / Allowlisting:** Restricts incoming traffic to specific IP addresses, ranges, or CIDR blocks.
    * **VPC / Network Peering:** Connects application cloud infrastructure directly to MongoDB Atlas environments via private cloud network peerings.
    * **Private Endpoints (PrivateLink):** Routes traffic over private endpoints (AWS PrivateLink, Azure Private Link, GCP Private Service Connect), completely isolating database traffic from the public internet.  
    ### Auditing & Log Redaction
    
    Provides accountability, operational visibility, and forensic tracking for regulatory compliance.  
    * **Database Auditing:** Captures administrative and data manipulation operations (e.g., authentication attempts, role modifications, index drops, schema updates) into structured JSON logs for audit trails.  
    * **Log Redaction:** Automatically strips sensitive data keys and clear-text field values from system log outputs to avoid accidental data leakage.

15. Explain the concept of embedded documents and when they should be used .

    An embedded document (or nested document) is a document stored inside another document as a field value in MongoDB's document-based model. Unlike relational databases that normalize data across separate tables connected by foreign keys, MongoDB allows related data to be stored together in a single BSON structure.

    #### Advantages :
    * **Single-Query Data Retrieval:** Related data is fetched in a single I/O operation without requiring multi-table JOIN operations or aggregation lookup stages.

    * **Atomic Updates:** Modifications to an embedded document and its parent occur within a single atomic operation, eliminating the need for multi-document ACID transactions for simple updates.

    * **Index Support:** You can create indexes on nested fields directly using dot notation (e.g., "metadata.version": 1).

    #### When to Use Embedded Documents
    * **One-to-One (1:1) Relationships:** When two entities are closely related and almost always queried together (e.g., a User and their Profile settings).

    * **One-to-Few (1:N) Relationships:** When a parent entity has a small, bounded set of child entities that belong exclusively to it (e.g., an Order with its line items, or a Contract with its signing history).

    * **Data is Queried Together:** When the application needs to read both parent and child data in the same view or screen nearly 100% of the time.

    * **Data Belongs Entirely to the Parent:** When child data has no independent lifecycle or meaning outside the parent entity and will never be shared across multiple parents.

16. What is the purpose of MongoDB’s $lookup stage in aggregation ?

    The $lookup stage in MongoDB’s aggregation pipeline performs an unsharded left outer join to combine fields from a separate target collection into the current pipeline documents.

    ### Primary Purposes
    * **Combine Data Across Collections:** Joins documents from two separate collections within the same database, allowing you to query normalized schemas.

    * **Denormalize on Read:** Retrieves related data at query time instead of duplicating data across documents during writes.

    * **Support Relational Patterns:** Emulates relational LEFT OUTER JOIN behavior, embedding matching documents into a specified array field.

18. What are the advantages of using MongoDB for horizontal scaling ?

    Horizontal scaling in MongoDB provides several key performance, operational, and cost advantages:
    * **Uncapped Storage & Throughput Limits:** Unlike vertical scaling (which is capped by the maximum RAM, CPU, and disk space of a single physical server), horizontal scaling allows clusters to expand capacity indefinitely by adding commodity server nodes.  
    * **Distributed Write Capacity:** In a standard replica set, a single primary node bears the write burden. Sharding assigns independent primary nodes to each shard, allowing write workloads to scale linearly across multiple servers.  
    * **Memory & Working Set Optimization:** By distributing a large collection across multiple shards, the active indexes and working dataset are split into smaller chunks. This allows more data to fit into system RAM across nodes, drastically reducing expensive disk I/O operations.

    * **Isolated Operational Failure:** If a single shard experiences localized hardware issues, only the data subset on that shard is impacted, while the remaining shards continue serving application traffic uninterrupted.  
    * **Geographic & Regulatory Compliance (Zoned Sharding):** Specific ranges of data can be assigned to designated shards located in target physical regions (e.g., placing European user records on EU-based shards to reduce latency and comply with data sovereignty regulations like GDPR).
    * **Cost Efficiency:** Adding standard, off-the-shelf instances or cloud nodes incrementally is significantly more cost-effective than provisioning ultra-high-tier enterprise hardware instances with extreme RAM or CPU specs.

19. How do MongoDB transactions differ from SQL transactions ?

    ### MongoDB Transactions

    * **Single-Document Atomicity:** Handles single-document operations natively with atomic locks, reducing the need for explicit transaction blocks in well-modeled embedded schemas.

    * **Isolation Model:** Applies Snapshot Isolation by default across all operations within a session snapshot.

    * **Conflict Resolution:** Rejects concurrent writes to the exact same document instantly, throwing a write conflict that requires the application layer to perform an explicit retry.

    * **Distributed Architecture:** Executes multi-document transactions across distributed sharded clusters using a native two-phase commit protocol managed by mongos routers.

    * **Design Philosophy:** Acts as an opt-in safety mechanism for multi-document operations, while encouraging embedded schemas to minimize transactional overhead.

    ### SQL Transactions

    * **Normalization Dependency:** Relies heavily on transaction blocks (BEGIN / COMMIT) for routine operations because normalized schemas split related data across multiple independent tables.

    * **Configurable Isolation:** Offers configurable isolation levels (Read Committed, Repeatable Read, Serializable) depending on application consistency requirements.

    * **Locking Mechanisms:** Employs row-level, predicate, or range-locking strategies (e.g., InnoDB, Postgres) that can cause concurrent writes to queue or block instead of immediately aborting.

    * **Scaling Focus:** Optimized primarily for single-node ACID guarantees, where cross-partition distributed transactions can introduce complex latency penalties.

    * **Design Philosophy:** Forms the core operational model, expecting applications to run multi-table joins and relational mutations within transaction boundaries by default.

20. What are the main differences between capped collections and regular collections ?

    ### Capped Collection

    * **Fixed Capacity Allocation:** Configured with a mandatory maximum size limit (in bytes) during creation, acting as a strict pre-allocated circular buffer.

    * **Automatic First-In, First-Out (FIFO) Eviction:** Overwrites the oldest documents automatically when the collection reaches its maximum size or document limit, eliminating manual cleanup.

    * **Preserved Insertion Order:** Guarantees that documents are stored and retrieved in their exact natural insertion order without requiring explicit $sort index stages.

    * **Restricted Document Operations:** Explicitly forbids individual document deletions (delete_one, delete_many) and restricts updates to operations that do not increase a document's original BSON size on disk.

    * **Tailable Cursor Support:** Supports long-polling tailable cursors (similar to tail -f) that remain open to stream newly inserted documents in real time.

    * **Sharding Limitations:** Cannot be sharded across a cluster, limiting its deployment to single instances or replica set nodes.

    ### Regular Collection

    * **Dynamic Growth:** Unbounded by default, automatically expanding to accommodate growing datasets as long as disk storage is available.

    * **Explicit Lifecycle Management:** Requires manual document deletions, collection drops, or automated Time-To-Live (TTL) indexes to expire and remove stale records.

    * **Arbitrary Update Flexibility:** Allows documents to expand dynamically during updates, shifting storage allocations automatically as fields are added.

    * **Full CRUD Lifecycle:** Fully supports standard single and multi-document deletions, updates, upserts, and complex field modifications.

    * **Full Sharding Capabilities:** Integrates seamlessly with MongoDB's sharding architecture to distribute data horizontally across multiple shards using a designated shard key.

    * **Change Stream Event System:** Relies on Change Streams via the replication oplog rather than tailable cursors to monitor real-time data modifications.

21. What is the purpose of the $match stage in MongoDB’s aggregation pipeline ?

    The $match stage filters documents in an aggregation pipeline so that only records matching specified search criteria are passed to subsequent pipeline stages. It functions identically to standard MongoDB query selectors used in find() queries.  
    ### Primary Purposes
    * **Document Filtering:** Filters out irrelevant records early, ensuring downstream pipeline stages (like $group or $sort) only process data that meets target conditions.  
    * **Index Utilization:** When placed as the first stage in an aggregation pipeline, $match can utilize collection indexes (like standard find() queries) to perform efficient index scans (IXSCAN) instead of full collection scans (COLLSCAN).
    * **Resource Optimization:** Reduces CPU, memory, and disk I/O overhead by minimizing the volume of BSON documents passed through subsequent pipeline steps.

    ### Identity Verification & Access Control
    * **Enable Authentication:** Enforce SCRAM-SHA-256 or x.509 certificates to ensure all clients and administrative tools verify identity before establishing a connection.

    * **Integrate Centralized Identity Providers:** Connect enterprise deployments to LDAP, Kerberos, or OIDC/OAuth 2.0 (e.g., Azure AD, Okta) for centralized user lifecycle management and Single Sign-On (SSO).

    * **Enforce Role-Based Access Control (RBAC):** Assign explicit roles (such as readWrite, dbAdmin) or construct custom roles to strictly limit operational privileges based on the principle of least privilege.

    ### Network Isolation & Boundary Protection
    * **Restrict IP Access:** Configure IP allowlists to limit cluster access exclusively to trusted application server IP ranges or specific subnet CIDR blocks.

    * **Use Private Cloud Endpoints:** Avoid binding MongoDB directly to public internet interfaces (0.0.0.0). Utilize VPC Peering or Private Service Endpoints (e.g., AWS PrivateLink, Azure Private Link) to route database traffic over isolated internal networks.

    * **Disable Default Network Bindings:** Ensure self-hosted mongod configurations explicitly set net.bindIp to internal localhost or private network interfaces.

    ### End-to-End Data Encryption
    * **In-Transit Encryption (TLS/SSL):** Enforce TLS 1.3 or TLS 1.2 across all client-to-database connections and internal replica set member communication.

    * **At-Rest Encryption:** Use WiredTiger engine encryption (AES-256) paired with key management services (AWS KMS, Azure Key Vault, or GCP KMS) to secure database files, logs, and backups on disk.

    * **In-Use Encryption (CSFLE & Queryable Encryption):** Implement Client-Side Field Level Encryption so sensitive fields (e.g., PII, API keys) are encrypted before leaving the application layer, ensuring data remains unreadable even to database administrators.

    ### System Auditing & Operational Governance
    * **Enable Activity Logging:** Capture key administrative activities, privilege changes, and authentication attempts into structured JSON audit logs for continuous monitoring.

    * **Redact Sensitive Logs:** Enable system log redaction to ensure clear-text field values and cryptographic tokens are stripped from log files prior to emission.

23. What is MongoDB’s WiredTiger storage engine, and why is it important ?

    WiredTiger is the default, multi-threaded storage engine for MongoDB responsible for managing how data is stored, read, and updated on disk and in physical RAM memory.

    ### Core Architecture & Mechanics

    * **Document-Level Concurrency Control:** Employs optimistic, document-level locking rather than collection-level or database-level locking. Multiple client threads can modify different documents within the same collection simultaneously without write contention.

    * **In-Memory Cache:** Allocates a dedicated RAM cache (typically 50% of available RAM minus 1 GB) to hold hot working sets, index pages, and recent uncommitted mutations before flushing them to disk.

    * **Checkpointing & Journaling:** Creates periodic checkpoints on disk (every 60 seconds or 2 GB of journal data) combined with write-ahead journaling (WiredTiger.wt) to ensure full durably crash recovery.

    * **B-Tree Data Structure:** Organizes collections and secondary indexes into optimized B-Tree structures on disk for efficient range scans and single-document point lookups.

    ### WiredTiger is Critical for Performance

    * **High Throughput Scaling:** Eliminates write bottlenecks in high-concurrency environments through fine-grained document locks.

    * **Storage Optimization (Snappy / Zlib Compression):** Compresses raw collections (up to 80% reduction) and index prefixes (up to 50% reduction) on disk by default, reducing storage infrastructure costs and accelerating disk I/O.

    * **At-Rest Security Integration:** Provides native AES-256 transparent encryption directly at the storage page level before data hits disk controllers.

    * **Snapshot Isolation:** Powers Multi-Version Concurrency Control (MVCC), providing read isolation and supporting MongoDB's multi-document ACID transactions.

```python
# 9.  Use aggregation to group orders by Region and calculate total sales per region .

from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")
db = client["superstore_db"]
collection = db["orders"]

pipeline = [
    {
        "$group": {
            "_id": "$Region",
            "total_sales": {"$sum": "$Sales"}
        }
    },
    {
        "$sort": {"total_sales": -1}
    }
]

results = collection.aggregate(pipeline)

for row in results:
    print(f"Region: {row['_id']:<10} | Total Sales: ${row['total_sales']:,.2f}")
```
